```python
# ==========================================
# CIFAR-100 Classification: Custom CNN + Transfer Learning
# ==========================================
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
from torchvision import datasets, transforms, models
from sklearn.metrics import classification_report
import numpy as np

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# -----------------------
# Data Loading & Augmentation
# -----------------------
mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)

train_tfms = transforms.Compose([
    transforms.RandomHorizontalFlip(),
    transforms.RandomCrop(32, padding=4),
    transforms.RandomRotation(15),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

test_tfms = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

full_train_dataset = datasets.CIFAR100(root="./data", train=True, download=True, transform=train_tfms)
test_dataset = datasets.CIFAR100(root="./data", train=False, download=True, transform=test_tfms)

train_size = int(0.8 * len(full_train_dataset))
val_size = len(full_train_dataset) - train_size
train_dataset, val_dataset = random_split(full_train_dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=128, shuffle=False, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=128, shuffle=False, num_workers=2)

# -----------------------
# Custom CNN
# -----------------------
class CustomCNN(nn.Module):
    def __init__(self, num_classes=100):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(),
            nn.Conv2d(64, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(64, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU(),
            nn.Conv2d(128, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(128, 256, 3, padding=1), nn.BatchNorm2d(256), nn.ReLU(),
            nn.Dropout(0.3),
            nn.MaxPool2d(2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256 * 4 * 4, 512), nn.ReLU(), nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

def top5_accuracy(output, target):
    with torch.no_grad():
        _, pred = output.topk(5, 1, True, True)
        correct5 = pred.eq(target.view(-1, 1).expand_as(pred))
        return correct5.sum().item()

def train_model(model, train_loader, val_loader, epochs=10, lr=1e-3):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    for epoch in range(epochs):
        model.train()
        running_loss = 0.0
        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()
        # Validation
        model.eval()
        correct1, correct5, total = 0, 0, 0
        with torch.no_grad():
            for images, labels in val_loader:
                images, labels = images.to(device), labels.to(device)
                outputs = model(images)
                _, predicted = outputs.max(1)
                total += labels.size(0)
                correct1 += predicted.eq(labels).sum().item()
                correct5 += top5_accuracy(outputs, labels)
        print(f"Epoch {epoch+1}/{epochs} "
              f"ValAcc: {100*correct1/total:.2f}% "
              f"ValTop5: {100*correct5/total:.2f}%")

def evaluate_model(model, loader):
    model.eval()
    correct1, correct5, total = 0, 0, 0
    all_labels, all_preds = [], []
    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct1 += predicted.eq(labels).sum().item()
            correct5 += top5_accuracy(outputs, labels)
            all_labels.extend(labels.cpu().numpy())
            all_preds.extend(predicted.cpu().numpy())
    print("Top-1 Accuracy:", 100*correct1/total, "%")
    print("Top-5 Accuracy:", 100*correct5/total, "%")
    print(classification_report(all_labels, all_preds, digits=3))

# Instantiate and train
custom_model = CustomCNN().to(device)
train_model(custom_model, train_loader, val_loader, epochs=20)  # increase epochs for better accuracy
print("Custom CNN on Test:")
evaluate_model(custom_model, test_loader)

# -----------------------
# Transfer Learning: EfficientNetB0
# -----------------------
transfer_model = models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.DEFAULT)
num_ftrs = transfer_model.classifier[1].in_features
transfer_model.classifier = nn.Sequential(
    nn.Dropout(p=0.3, inplace=True),
    nn.Linear(num_ftrs, 100)
)
transfer_model = transfer_model.to(device)

# Stage 1: freeze backbone
for param in transfer_model.features.parameters():
    param.requires_grad = False

train_model(transfer_model, train_loader, val_loader, epochs=5, lr=1e-3)  # train head
print("Transfer Learning Stage 1 Test:")
evaluate_model(transfer_model, test_loader)

# Stage 2: unfreeze top layers
for param in transfer_model.features[-2:].parameters():
    param.requires_grad = True

train_model(transfer_model, train_loader, val_loader, epochs=5, lr=1e-4)  # fine-tune
print("Transfer Learning Stage 2 Test:")
evaluate_model(transfer_model, test_loader)
```
